```python
import json
```

```python
base = "../eval_result/benchmark_base_comp_datagen.json"
with open(base, "r", encoding="utf-8") as f:
    base_data = json.load(f)
```

```python
tutor = "../eval_result/benchmark_tutor_prompt_datagen.json"
with open(tutor, "r", encoding="utf-8") as f:
    tutor_data = json.load(f)
```

```python
for b_data, t_data in zip(base_data, tutor_data):
    print("=>",b_data['category'])
    print("=>",b_data['query'])
    print("=>",b_data['ans_gpt'][:100])
    print("=>",b_data['ans_gemini'][:100])
    print("=>",b_data['ans_ours'][:100])
    print("=>",t_data['ans_gpt'][:100])
    print("=>",t_data['ans_gemini'][:100])
    break
```

```python
out = []
for b, t in zip(base_data, tutor_data):
    out.append({
        "category": b.get("category"),
        "query": b.get("query"),
        'gpt_raw': b.get("ans_gpt"),
        'gemini_raw': b.get("ans_gemini"),
        'gpt_tutor': t.get("ans_gpt"),
        'gemini_tutor': t.get("ans_gemini"),
        'ours': b.get("ans_ours"),
    })

out_fname = "../eval_result/combined_base_tutor.json"
with open(out_fname, "w", encoding="utf-8") as f:
    json.dump(out, f, ensure_ascii=False, indent=2)

print("Saved:", out_fname)
```
