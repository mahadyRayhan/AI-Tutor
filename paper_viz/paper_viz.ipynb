```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Journal Quality Settings ---
# Set global font properties for a professional look
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 18,
    'font.weight': 'bold',
    'axes.labelsize': 28,
    'axes.titlesize': 32,
    'xtick.labelsize': 30,
    'ytick.labelsize': 30,
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold'
})

# Set Seaborn style for a clean grid with dashed lines
sns.set_style("whitegrid", {'grid.linestyle': '--'})

# --- Data ---
data = {
    'Category': [
        'Output Safety',
        'Privacy',
        'Infrastructure',
        'Tool Abuse',
        'RAG Leakage / Integrity',
        'Jailbreak',
        'Indirect Poisoning',
        'Social Engineering',
        'RBAC / Data Leakage'
    ],
    'Block_Rate': [
        75,
        100,
        100,
        100,
        100,
        100,
        100,
        100,
        80
    ],
    'Annotation': [
        '75% (3/4)',
        '',
        '',
        '',
        '',
        '',
        '',
        '',
        '80% (4/5)'
    ]
}

df = pd.DataFrame(data)

# --- Plotting ---
# INCREASED WIDTH: (16, 10) ensures long labels on the left fit comfortably
plt.figure(figsize=(16, 10))

# Use a single, solid color
color = sns.color_palette("tab10")[0]

# Create horizontal bar plot
bars = plt.barh(df['Category'], df['Block_Rate'], color=color)

# Set labels
plt.xlabel('Block Rate (%)')

# Add annotations
for bar, annotation in zip(bars, df['Annotation']):
    if annotation:
        plt.text(
            bar.get_width() + 1, 
            bar.get_y() + bar.get_height() / 2,
            annotation,
            va='center',
            ha='left',
            fontsize=30,
            fontweight='bold'
        )

# Set x-axis limits and invert y-axis
plt.xlim(0, 100)
plt.gca().invert_yaxis()

# Remove top and right spines
sns.despine()

# AUTO-ADJUST LAYOUT: This is crucial for fixing cut-off labels
plt.tight_layout()

# --- Save the figure ---
# bbox_inches='tight' trims whitespace, pad_inches adds a safety margin
plt.savefig('security_block_rates.png', dpi=600, bbox_inches='tight', pad_inches=0.1)
plt.savefig('security_block_rates.pdf', format='pdf', bbox_inches='tight', pad_inches=0.1)

print("Saved 'security_block_rates.png' (High-res PNG)")
print("Saved 'security_block_rates.pdf' (Vector PDF)")
# Reset defaults
plt.rcParams.update(plt.rcParamsDefault)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- Journal Quality Settings ---
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 18,
    'font.weight': 'bold',
    'axes.labelsize': 24,
    'axes.titlesize': 28,
    'xtick.labelsize': 20, # Slightly smaller to fit the extra text
    'ytick.labelsize': 20,
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold',
    'legend.fontsize': 18
})

sns.set_style("whitegrid", {'grid.linestyle': '--'})

# --- Data ---
categories = ['Code Density', 'Security Failure', 'Scaffolding']

# UPDATED LABELS: explicit directionality included
categories_with_dir = [
    'Code Density\n(Lower is better $\downarrow$)', 
    'Security Failure\n(Lower is better $\downarrow$)', 
    'Scaffolding\n(Higher is better $\\uparrow$)'
]

gpt5_vals = [68.4, 80, 5]
gemini3_vals = [55.2, 90, 10]
dpac_vals = [12.1, 0, 95]

# --- Plotting ---
plt.figure(figsize=(14, 11)) # Increased height slightly for multi-line x-labels

bar_width = 0.2
r1 = np.arange(len(categories))
r2 = [x + bar_width for x in r1]
r3 = [x + bar_width for x in r2]

# Colors
c1_face, c1_edge = '#a6a6ff', 'blue'
c2_face, c2_edge = '#ff9999', '#cc0000'
c3_face, c3_edge = '#e6dcc3', '#8b5a2b'

# Create Bars
bars1 = plt.bar(r1, gpt5_vals, color=c1_face, width=bar_width, edgecolor=c1_edge, linewidth=1.5, label='GPT-5')
bars2 = plt.bar(r2, gemini3_vals, color=c2_face, width=bar_width, edgecolor=c2_edge, linewidth=1.5, label='Gemini 3')
bars3 = plt.bar(r3, dpac_vals, color=c3_face, width=bar_width, edgecolor=c3_edge, linewidth=1.5, label='DPAC-RAG')

# Add Labels on bars
def add_labels(bars, color, vals):
    for bar, val in zip(bars, vals):
        height = bar.get_height()
        y_pos = height + 1.5 if height > 0 else 1.5
        plt.text(
            bar.get_x() + bar.get_width() / 2, 
            y_pos,
            f'{val}', 
            ha='center', va='bottom', color=color, fontsize=18, fontweight='bold'
        )

add_labels(bars1, c1_edge, gpt5_vals)
add_labels(bars2, c2_edge, gemini3_vals)
add_labels(bars3, c3_edge, dpac_vals)

# Formatting
plt.ylabel('Percentage (%)')

# Apply the new labels with directionality
plt.xticks([r + bar_width for r in range(len(categories))], categories_with_dir)
plt.ylim(0, 115) 

sns.despine()

# Legend
plt.legend(
    loc='upper center', 
    bbox_to_anchor=(0.5, -0.15), # Moved down further to avoid overlap with multi-line x-labels
    ncol=3, 
    frameon=True, 
    edgecolor='black', 
    fancybox=False
)

plt.tight_layout()

# --- Save ---
plt.savefig('grouped_barchart_directionality.png', dpi=600, bbox_inches='tight')
plt.savefig('grouped_barchart_directionality.pdf', format='pdf', bbox_inches='tight')

print("Saved 'grouped_barchart_directionality.png'")
print("Saved 'grouped_barchart_directionality.pdf'")

# Reset defaults
plt.rcParams.update(plt.rcParamsDefault)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# --- Journal Quality Settings ---
# Increased font sizes further for better visibility
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 20,         # Global font size
    'font.weight': 'bold',
    'axes.labelsize': 36,    # Axis Labels
    'axes.titlesize': 40,    # Title
    'xtick.labelsize': 38,   # X Ticks
    'ytick.labelsize': 38,   # Y Ticks
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold',
})

# Set Seaborn style
sns.set_style("whitegrid", {'grid.linestyle': '--'})

# =========================
# Data
# =========================
systems = ["GPT Raw", "GPT Tutor", "Gemini Raw", "Gemini Tutor", "DPAC-RAG"]

# X-axis: Overall Judge Preference / Win Rate (%)
preference = [4.0, 42.0, 6.0, 4.0, 44.0]

# Components of compliance (%)
security_compliance = [20.0, 30.0, 20.0, 30.0, 90.0]
curriculum_compliance = [0.0, 0.0, 0.0, 0.0, 80.0]

# Y-axis: Combined compliance score
combined_compliance = [
    (s + c) / 2.0 for s, c in zip(security_compliance, curriculum_compliance)
]

# =========================
# Plot
# =========================
plt.figure(figsize=(16, 11))

colors = sns.color_palette("deep", n_colors=len(systems))

# Plot Scatter with larger markers
plt.scatter(
    preference, 
    combined_compliance, 
    s=500, 
    c=colors, 
    edgecolor='black', 
    linewidth=2, 
    zorder=3,
    alpha=0.9
)

# --- Manual Label Adjustment to Fix Overlaps ---
# Dictionary defining (x_offset, y_offset) for each system
offsets = {
    "GPT Raw":      (-0.5, -5.0),  # Move label BELOW the dot
    "Gemini Raw":   (1.0, -1.0),   # Move label to the RIGHT
    "Gemini Tutor": (0.5, 3.5),   # Move label ABOVE the dot
    "GPT Tutor":    (1.0, 0.0),    # Move label to the RIGHT
    "DPAC-RAG":     (1.0, 3.5)    # Move label ABOVE-LEFT
}

# Horizontal alignment for cleaner positioning
alignments = {
    "GPT Raw":      'center',
    "Gemini Raw":   'left',
    "Gemini Tutor": 'center',
    "GPT Tutor":    'left',
    "DPAC-RAG":     'center' 
}

for name, x, y in zip(systems, preference, combined_compliance):
    x_off, y_off = offsets.get(name, (1.5, 1.0))
    ha = alignments.get(name, 'left')
    
    plt.text(
        x + x_off, 
        y + y_off, 
        name, 
        fontsize=35,         # Big, bold text
        fontweight='bold',
        va='center',
        ha=ha
    )

plt.xlabel("Overall Judge Preference (Win Rate, %)")
plt.ylabel("Combined Compliance Score (%)\n(avg. of Security + Curriculum)")
# plt.title("Trade-off Between Preference and Compliance")

# Set bounds (Start X at -2 to accommodate left-aligned labels)
plt.xlim(-2, 55)
plt.ylim(0, 100)

sns.despine()
plt.tight_layout()

# Save high-res
plt.savefig("tradeoff_preference_vs_compliance_journal.png", dpi=600, bbox_inches='tight')
plt.savefig("tradeoff_preference_vs_compliance_journal.pdf", format='pdf', bbox_inches='tight')

print("Saved 'tradeoff_preference_vs_compliance_journal.png'")
print("Saved 'tradeoff_preference_vs_compliance_journal.pdf'")

# Reset defaults
plt.rcParams.update(plt.rcParamsDefault)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- Journal Quality Settings ---
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 20,
    'font.weight': 'bold',
    'axes.labelsize': 36,
    'axes.titlesize': 40,
    'xtick.labelsize': 22,
    'ytick.labelsize': 40,
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold',
    'legend.fontsize': 30,
    'hatch.linewidth': 2.0  # Increased to make the texture clear
})

sns.set_style("whitegrid", {'grid.linestyle': '--'})

# --- Data ---
categories = ['Pedagogy', 'Security', 'Boundary', 'OVERALL']
categories_labels = [
    'Pedagogy\n(Lower is better $\\downarrow$)', 
    'Security\n(Lower is better $\\downarrow$)', 
    'Boundary\n(Lower is better $\\downarrow$)', 
    'OVERALL\n(Lower is better $\\downarrow$)'
]

gpt5_vals = [91.7, 22.2, 88.9, 70.0]
dpac_vals = [25.0, 0.0, 16.7, 15.0]

# --- Plotting ---
plt.figure(figsize=(14, 11))

bar_width = 0.35
r1 = np.arange(len(categories))
r2 = [x + bar_width for x in r1]

# --- COLORS & TEXTURES (Replicating the Example) ---

# Style 1 (like the "RMSE" bars in your image): Diagonal Hatch
# Light Orange fill with Dark Red/Orange edge
c1_face = '#ffcc99'  
c1_edge = '#cc3300'
hatch1 = '//'

# Style 2 (like the "MAE" bars in your image): Dotted Hatch
# Light Green fill with Dark Green edge
c2_face = '#99ffcc'
c2_edge = '#006633'
hatch2 = '..'

# Create Bars with Hatching
bars1 = plt.bar(r1, gpt5_vals, color=c1_face, width=bar_width, 
                edgecolor=c1_edge, linewidth=2, hatch=hatch1, 
                label='GPT-5 (Tutor Mode)')

bars2 = plt.bar(r2, dpac_vals, color=c2_face, width=bar_width, 
                edgecolor=c2_edge, linewidth=2, hatch=hatch2, 
                label='DPAC-RAG (Ours)')

# Add Value Labels
def add_labels(bars, color, vals):
    for bar, val in zip(bars, vals):
        height = bar.get_height()
        y_pos = height + 1.5 if height > 0 else 1.5
        
        plt.text(
            bar.get_x() + bar.get_width() / 2 + 0.05, 
            y_pos,
            f'{val:.1f}%', 
            ha='center', 
            va='bottom', 
            color=color, 
            fontsize=30, 
            fontweight='bold'
        )

add_labels(bars1, c1_edge, gpt5_vals)
add_labels(bars2, c2_edge, dpac_vals)

# Formatting
plt.ylabel('Violation / Failure Rate (%)')
plt.xticks([r + bar_width/2 for r in range(len(categories))], categories_labels)
plt.ylim(0, 110)

sns.despine()

# Legend
plt.legend(
    loc='upper center', 
    bbox_to_anchor=(0.5, -0.15), 
    ncol=2, 
    frameon=True, 
    edgecolor='black', 
    fancybox=False
)

plt.tight_layout()

# --- Save ---
plt.savefig('solution_exposure_rate.png', dpi=600, bbox_inches='tight')
plt.savefig('solution_exposure_rate.pdf', format='pdf', bbox_inches='tight')

print("Saved 'solution_exposure_rate.png'")
print("Saved 'solution_exposure_rate.pdf'")

# Reset defaults
plt.rcParams.update(plt.rcParamsDefault)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Journal Quality Settings ---
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 25,
    'font.weight': 'bold',
    'axes.labelsize': 40,
    'axes.titlesize': 36,   # Large Title
    'xtick.labelsize': 34,
    'ytick.labelsize': 40,
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold',
    'hatch.linewidth': 2.0  # Thicker texture patterns
})

# Set Seaborn style
sns.set_style("whitegrid", {'grid.linestyle': '--'})

# --- Data ---
data = {
    'Category': [
        'Output Safety',
        'Privacy',
        'Infrastructure',
        'Tool Abuse',
        'RAG Leakage / Integrity',
        'Jailbreak',
        'Indirect Poisoning',
        'Social Engineering',
        'RBAC / Data Leakage'
    ],
    'Block_Rate': [
        75,
        100,
        100,
        100,
        100,
        100,
        100,
        100,
        80
    ],
    'Annotation': [
        '75% (3/4)',
        '100%',
        '100%',
        '100%',
        '100%',
        '100%',
        '100%',
        '100%',
        '80% (4/5)'
    ]
}

df = pd.DataFrame(data)

# --- Visual Logic: Colors & Textures ---
colors = []
hatches = []
edge_colors = []

for rate in df['Block_Rate']:
    if rate == 100:
        # Perfect Score: Emerald Green with Dots
        colors.append('#a8e6cf')   # Light Green Fill
        edge_colors.append('#1b5e20') # Dark Green Edge
        hatches.append('..')       # Dot texture
    else:
        # Warning Score: Salmon/Red with Stripes
        colors.append('#ff8b94')   # Light Red Fill
        edge_colors.append('#b71c1c') # Dark Red Edge
        hatches.append('//')       # Stripe texture

# --- Plotting ---
plt.figure(figsize=(16, 11)) # Increased height slightly for title room

# Draw horizontal bars
bars = plt.barh(df['Category'], df['Block_Rate'], color=colors, edgecolor=edge_colors, linewidth=2, height=0.7)

# Apply hatches individually
for bar, hatch in zip(bars, hatches):
    bar.set_hatch(hatch)

# Set labels and title
plt.xlabel('Block Rate (%)')
# plt.title('Security Block Rates by Category') # Added a title to utilize the space

# Add annotations
for bar, annotation in zip(bars, df['Annotation']):
    width = bar.get_width()
    plt.text(
        width + 1, 
        bar.get_y() + bar.get_height() / 2,
        annotation,
        va='center',
        ha='left',
        fontsize=30,
        fontweight='bold',
        color='black'
    )

# Formatting
plt.xlim(0, 112) 
plt.gca().invert_yaxis() 

# --- Force Bold Tick Labels ---
# Sometimes rcParams doesn't catch everything, this forces it:
plt.xticks(fontweight='bold')
plt.yticks(fontweight='bold')

# --- Aesthetic Fixes ---
sns.despine(left=True, bottom=False)

# Grid lines
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.grid(axis='y', visible=False) 

# --- LAYOUT FIX for Title Cut-off ---
# rect=[left, bottom, right, top]
# We limit the plot to the bottom 92%, leaving top 8% for the massive title
# plt.tight_layout(rect=[0, 0, 1, 0.92])

# --- Save ---
plt.savefig('security_block_rates_vibrant.png', dpi=600, bbox_inches='tight', pad_inches=0.3)
plt.savefig('security_block_rates_vibrant.pdf', format='pdf', bbox_inches='tight', pad_inches=0.3)

print("Saved 'security_block_rates_vibrant.png' (High-res PNG)")
print("Saved 'security_block_rates_vibrant.pdf' (Vector PDF)")

# Reset defaults
plt.rcParams.update(plt.rcParamsDefault)
```
