```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- Journal Quality Settings ---
# Set global font properties for a professional look
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 18,
    'font.weight': 'bold',
    'axes.labelsize': 24,
    'axes.titlesize': 28,
    'xtick.labelsize': 20,
    'ytick.labelsize': 20,
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold'
})

# Set Seaborn style for a clean grid with dashed lines
sns.set_style("whitegrid", {'grid.linestyle': '--'})

# --- Data ---
data = {
    'Category': [
        'Output Safety',
        'Privacy',
        'Infrastructure',
        'Tool Abuse',
        'RAG Leakage / Integrity',
        'Jailbreak',
        'Indirect Poisoning',
        'Social Engineering',
        'RBAC / Data Leakage'
    ],
    'Block_Rate': [
        75,
        100,
        100,
        100,
        100,
        100,
        100,
        100,
        80
    ],
    'Annotation': [
        '75% (3/4)',
        '',
        '',
        '',
        '',
        '',
        '',
        '',
        '80% (4/5)'
    ]
}

df = pd.DataFrame(data)

# --- Plotting ---
# INCREASED WIDTH: (16, 10) ensures long labels on the left fit comfortably
plt.figure(figsize=(16, 10))

# Use a single, solid color
color = sns.color_palette("tab10")[0]

# Create horizontal bar plot
bars = plt.barh(df['Category'], df['Block_Rate'], color=color)

# Set labels
plt.xlabel('Block Rate (%)')

# Add annotations
for bar, annotation in zip(bars, df['Annotation']):
    if annotation:
        plt.text(
            bar.get_width() + 1, 
            bar.get_y() + bar.get_height() / 2,
            annotation,
            va='center',
            ha='left',
            fontsize=22,
            fontweight='bold'
        )

# Set x-axis limits and invert y-axis
plt.xlim(0, 100)
plt.gca().invert_yaxis()

# Remove top and right spines
sns.despine()

# AUTO-ADJUST LAYOUT: This is crucial for fixing cut-off labels
plt.tight_layout()

# --- Save the figure ---
# bbox_inches='tight' trims whitespace, pad_inches adds a safety margin
plt.savefig('security_block_rates.png', dpi=600, bbox_inches='tight', pad_inches=0.1)
plt.savefig('security_block_rates.pdf', format='pdf', bbox_inches='tight', pad_inches=0.1)

print("Saved 'security_block_rates.png' (High-res PNG)")
print("Saved 'security_block_rates.pdf' (Vector PDF)")
# Reset defaults
plt.rcParams.update(plt.rcParamsDefault)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- Journal Quality Settings ---
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 18,
    'font.weight': 'bold',
    'axes.labelsize': 24,
    'axes.titlesize': 28,
    'xtick.labelsize': 20, # Slightly smaller to fit the extra text
    'ytick.labelsize': 20,
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold',
    'legend.fontsize': 18
})

sns.set_style("whitegrid", {'grid.linestyle': '--'})

# --- Data ---
categories = ['Code Density', 'Security Failure', 'Scaffolding']

# UPDATED LABELS: explicit directionality included
categories_with_dir = [
    'Code Density\n(Lower is better $\downarrow$)', 
    'Security Failure\n(Lower is better $\downarrow$)', 
    'Scaffolding\n(Higher is better $\\uparrow$)'
]

gpt5_vals = [68.4, 80, 5]
gemini3_vals = [55.2, 90, 10]
dpac_vals = [12.1, 0, 95]

# --- Plotting ---
plt.figure(figsize=(14, 11)) # Increased height slightly for multi-line x-labels

bar_width = 0.2
r1 = np.arange(len(categories))
r2 = [x + bar_width for x in r1]
r3 = [x + bar_width for x in r2]

# Colors
c1_face, c1_edge = '#a6a6ff', 'blue'
c2_face, c2_edge = '#ff9999', '#cc0000'
c3_face, c3_edge = '#e6dcc3', '#8b5a2b'

# Create Bars
bars1 = plt.bar(r1, gpt5_vals, color=c1_face, width=bar_width, edgecolor=c1_edge, linewidth=1.5, label='GPT-5')
bars2 = plt.bar(r2, gemini3_vals, color=c2_face, width=bar_width, edgecolor=c2_edge, linewidth=1.5, label='Gemini 3')
bars3 = plt.bar(r3, dpac_vals, color=c3_face, width=bar_width, edgecolor=c3_edge, linewidth=1.5, label='DPAC-RAG')

# Add Labels on bars
def add_labels(bars, color, vals):
    for bar, val in zip(bars, vals):
        height = bar.get_height()
        y_pos = height + 1.5 if height > 0 else 1.5
        plt.text(
            bar.get_x() + bar.get_width() / 2, 
            y_pos,
            f'{val}', 
            ha='center', va='bottom', color=color, fontsize=18, fontweight='bold'
        )

add_labels(bars1, c1_edge, gpt5_vals)
add_labels(bars2, c2_edge, gemini3_vals)
add_labels(bars3, c3_edge, dpac_vals)

# Formatting
plt.ylabel('Percentage (%)')

# Apply the new labels with directionality
plt.xticks([r + bar_width for r in range(len(categories))], categories_with_dir)
plt.ylim(0, 115) 

sns.despine()

# Legend
plt.legend(
    loc='upper center', 
    bbox_to_anchor=(0.5, -0.15), # Moved down further to avoid overlap with multi-line x-labels
    ncol=3, 
    frameon=True, 
    edgecolor='black', 
    fancybox=False
)

plt.tight_layout()

# --- Save ---
plt.savefig('grouped_barchart_directionality.png', dpi=600, bbox_inches='tight')
plt.savefig('grouped_barchart_directionality.pdf', format='pdf', bbox_inches='tight')

print("Saved 'grouped_barchart_directionality.png'")
print("Saved 'grouped_barchart_directionality.pdf'")

# Reset defaults
plt.rcParams.update(plt.rcParamsDefault)
```
